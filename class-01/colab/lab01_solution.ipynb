# Lab 1: Photo → Clean Edge Sketch
Load a photo, crop the main subject, and turn it into a black-and-white line drawing.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def show(*images, titles=None, size=5):
    """Show one or more images side by side (handles BGR colour and grayscale)."""
    titles = titles or [""] * len(images)
    plt.figure(figsize=(size * len(images), size))
    for k, (im, t) in enumerate(zip(images, titles), 1):
        plt.subplot(1, len(images), k)
        if im.ndim == 2:
            plt.imshow(im, cmap="gray", vmin=0, vmax=255)
        else:
            plt.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB))
        plt.title(t); plt.axis("off")
    plt.show()
```

```python
img = cv2.imread(path)
assert img is not None, "Could not read the image. Check the file name!"
print("shape:", img.shape, " dtype:", img.dtype)
show(img, titles=["original"])
```

```python
h, w = img.shape[:2]
roi = img[int(h * 0.10):int(h * 0.90), int(w * 0.20):int(w * 0.80)]

gray   = cv2.cvtColor(roi, cv2.COLOR_BGR2GRAY)
blur   = cv2.GaussianBlur(gray, (5, 5), 0)
edges  = cv2.Canny(blur, 50, 150)
edges  = cv2.dilate(edges, np.ones((2, 2), np.uint8))   # bonus: thicker lines
sketch = 255 - edges

show(roi, sketch, titles=["crop", "sketch"])
cv2.imwrite("sketch.png", sketch)
```
